By using the info() method, I can gather more information about the data. 
The data has 20,000 rows and 18 columns. There is one categorical column which is gender. It also shows the non-null values. I can confirm that there are no non-null values which means that there are no missing values in my dataset.

```python
# using data types to check the data types of the columns
df.dtypes
```

```python
# show unique values in the gender column
df['gender'].unique()
```

```python
# show the count of each unique value in the gender column
df['gender'].value_counts()
```

Using the methods above shows a little more depth analysis for the categorical variable in my dataset.

```python
# using describe() to get a statistical summary of the dataset
df.describe()
```

The describe() summarises the central tendency, dispersion and shape of a dataset. This is useful as for numerical data shows the count, mean, std, min, max, lower, 50 and upper percentiles. This can help identify outliers.

```python
# using corr() to check the correlation between the numerical columns only in the dataset
df.select_dtypes(include=np.number).corr()
```

 I used the .select_dtypes to include the the numerical data only.

```python
# using spearman method to check the correlation between the numerical columns only in the dataset
df.select_dtypes(include=np.number).corr(method='spearman')
```

Using the 'spearman' method, this is to assess the monotonic relationships in numerical variables.
