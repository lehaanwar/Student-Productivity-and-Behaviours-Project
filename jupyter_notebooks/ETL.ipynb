```python
# check for outliers in the numerical columns using boxplot
numeric_df = df.select_dtypes(include=np.number)
plt.figure(figsize=(10, 6))
numeric_df.boxplot(rot=90)
plt.title('Boxplot of Numerical Variables')
plt.show()
```

The above boxplot looks very strange because the student_id has a large range from 0-20,000, whilst the other columns are squashed and much smaller. I will drop the student_id as it does not carry any statistical meaning. 

```python
# drop the student_id column as it is not useful for analysis. Use the boxplot to check for outliers in the numerical columns again after dropping the student_id column
df.drop('student_id', axis=1, inplace=True) 
numeric_df = df.select_dtypes(include=np.number)
plt.figure(figsize=(10, 6))
numeric_df.boxplot(rot=90)
plt.title('Boxplot of Numerical Variables after dropping student_id column')
plt.show()
```

From the above boxplot, there are two columns that stand out. One is coffee_intake_mg and the other is productivity_score. The coffee_intake_mg is the largest, this could just mean that the student consume coffee is very different amounts. As there is no dots outside the whiskers, there are no extreme outliers. However, there are two dots outside the productivity_score, 0 and 100. we will look at this is in more detail.

```python
# using scatterplot to show outliers in productivity_score column only  
plt.figure(figsize=(10, 6))
sns.scatterplot(x=df.index, y='productivity_score', data=df)    
plt.title('Scatterplot of Productivity Score')
plt.xlabel('Index')
plt.ylabel('Productivity Score')
plt.show()
```

By creating the above plot for productivity_score, it shows there are more than one student that received 0 and 100. I will analyse any factors that could cause a student to score 0 and 100.

```python
# check how many students have a productivity score of 100 and how many have a productivity score of 0
score_0 = df[df['productivity_score'] == 0].shape[0]
score_100 = df[df['productivity_score'] == 100].shape[0]
print("Number of students with a productivity score of 0:", score_0)
print("Number of students with a productivity score of 100:", score_100)
```

```python
# using .loc to find the outliers in productivity_score
df.loc[(df['productivity_score'] == 0) | (df['productivity_score'] == 100)]
```

As the above shows that the student who scored 0 had low attendance, less sleep, less study hours per day, higher stress level and lower final grade. Which is the complete opposite of the student who had a productivity score of 100. I decided to keep these rather than remove.

Create a new transformer in ML feature engineering

```python
# create a new feature called 'screen_time', this will be the sum of the time spent on phone, youtube, gaming and social media.
df['screen_time'] = df['phone_usage_hours'] + df['youtube_hours'] + df['gaming_hours'] + df['social_media_hours']
df.head()
```

```python
#create a new feature called 'wellness_score', this will be the sum of the time spent on sleep, exercise.
df['wellness_score'] = df['sleep_hours'] + df['exercise_minutes'] / 60
df.head()
```

New features created successfully
