Core concepts such as mean, median, standard deviation, hypothesis testing, and basic probability. These will be outlined below.

Mean is the average of a set of numbers. The mean is a very important measure of central tendency, which represents the typical value of a dataset. The mean provides a summary of the entire data making it easier to understand the overall patterns of the dataset. I will use the mean to help understand the average of a student's study_hours_per_day, assignments_completed and final_grade

```python
# calculate the mean study hours per day, attendance percentage, final grade.
mean_study_hours = df['study_hours_per_day'].mean()
mean_attendance_percentage = df['attendance_percentage'].mean()
mean_final_grade = df['final_grade'].mean()
print("Mean study hours per day:", mean_study_hours)
print("Mean attendance percentage:", mean_attendance_percentage)
print("Mean final grade:", mean_final_grade)
```

Median is the middle value of the data in an ascending or descending order. The median is known as the central value of any given data. 

```python
#median of study hours per day, attendance percentage, final grade.
median_study_hours = df['study_hours_per_day'].median()
median_attendance_percentage = df['attendance_percentage'].median()
median_final_grade = df['final_grade'].median()
print("Median study hours per day:", median_study_hours)
print("Median attendance percentage:", median_attendance_percentage)
print("Median final grade:", median_final_grade)
```

Standard deviation describes how spread out the values are around the mean. There is a high standard deviation and low standard deviation. A low standard deviation shows that most of the numbers are close to the mean. A high standard deviation means that the value are spread out over a wider range. This can provide how consistent student behaviour.

```python
# standard deviation of study hours per day, attendance percentage, final grade.
std_study_hours = df['study_hours_per_day'].std()
std_attendance_percentage = df['attendance_percentage'].std()
std_final_grade = df['final_grade'].std()
print("Standard deviation of study hours per day:", std_study_hours)
print("Standard deviation of attendance percentage:", std_attendance_percentage)
print("Standard deviation of final grade:", std_final_grade)
```

Hypothesis testing compares two conflicting hypothesis. the key concept is Null Hypothesis which assumes that there is no effect. It assumes a given statement is true unless it is proven false.
The Alternative Hypothesis which shows that there is a difference or a significant effect. It raises doubt that the given statement can be false.
P-value is to check if the data of the null hypothesis is true. If it is under 0.05 then the null hypothesis is false.
T-test is used to compare the averages of two groups to see if they are significantly different from each other. We will use this to compare the mean final grade of students of group 1 who study more than 4 hours per day and group 2 who study less than 4 hours per day

Null Hypothesis: THere is no difference in the final grades of students who study less than four hours per day than those who study more than four hours.

Alternate Hypotheis: Students who study more than four hours per day will have a have a higher final grade.

```python
#hypothesis testing using t-test to compare the mean final grade of students who study more than 4 hours per day and those who study less than 4 hours per day.
from scipy.stats import ttest_ind
group1 = df[df['study_hours_per_day'] > 4]['final_grade']
group2 = df[df['study_hours_per_day'] <= 4]['final_grade']
t_stat, p_value = ttest_ind(group1, group2)
print("T-statistic:", t_stat)
print("P-value:", p_value)
```

The p-value is 0.31 which is greater than 0.05, I fail to reject the null hypothesis. The data does not provide enough evidence.

Basic Probability confirms the likelihood of something is to happen. It gives a numerical value to help solve uncertainty in a dataset. It is expressed as a number between 0 and 1. 0 meaning impossible and 1 is certain.

```python
#probability of students who study more than 4 hours per day getting a final grade above 85.
probability = len(df[(df['study_hours_per_day'] > 4) & (df['final_grade'] > 85)]) / len(df[df['study_hours_per_day'] > 4])
print("Probability of students who study more than 4 hours per day getting a final grade above 85:", probability)
```

Machine Learning Algorithm - K-means Clustering

I will examine k-means clustering to identify student behaviour groups and see which clusters have high or low productivity scores. Check if there are any patterns between bahaviour and productivity. I will not be using alll the variables for this. Only the variables that describe behaviour: study_hours_per_day,sleep_hours,phone_usage_hours, social_media_hours	youtube_hours, gaming_hours, breaks_per_day, exercise_minutes and stress_level.


```python
# import libraries for K-means clustering
from sklearn.cluster import KMeans
from sklearn.preprocessing import StandardScaler
```

```python
#selecting features for clustering
features = df[['study_hours_per_day', 'social_media_hours', 'youtube_hours', 'sleep_hours', 'phone_usage_hours', 'gaming_hours', 'stress_level','breaks_per_day','exercise_minutes']]
```

```python
# scaling the features
scaler = StandardScaler()
scaled_features = scaler.fit_transform(features)
```

```python
#calculating the optimal number of clusters using the elbow method
inertia = []
for k in range(1, 11):
    kmeans = KMeans(n_clusters=k, random_state=42)
    kmeans.fit(scaled_features)
    inertia.append(kmeans.inertia_)

# plotting the elbow curve
plt.figure(figsize=(10, 6))
plt.plot(range(1, 11), inertia, marker='o')
plt.title('Elbow Method for Optimal k')
plt.xlabel('Number of clusters (k)')
plt.ylabel('Inertia')
plt.xticks(range(1, 11))
plt.show()
```

The elbow method is used to see that the optimal number of clusters. From the above, we can see the inertia decrease around four clusters, so four clusters will be chosen as the optimal number of clusters.

```python
# labelling the data with the optimal number of clusters k=4
kmeans = KMeans(n_clusters=4, random_state=42)
kmeans.fit(scaled_features)
df['cluster'] = kmeans.labels_
```

```python
# groupby cluster to find the average of study hours per day, social media hours, youtube hours, sleep hours, phone usage hours, gaming hours, stress level, breaks per day and exercise minutes for each cluster.
cluster_summary = df.groupby('cluster')[['study_hours_per_day', 'social_media_hours', 'youtube_hours', 'sleep_hours', 'phone_usage_hours', 'gaming_hours', 'stress_level','breaks_per_day','exercise_minutes']].mean()
print(cluster_summary)
```

```python
# choose cluster names based on the above summary
cluster_names= {0:'low_productivity', 1: 'moderate_productivity', 2: 'high_productivity', 3: 'very_high_productivity'}
df['cluster_label'] = df['cluster'].map(cluster_names)
print(df[['cluster', 'cluster_label']].value_counts())

```

```python
# visualise clusters using a scatter plot of study hours per day and productivity score, colored by cluster name.
plt.figure(figsize=(10, 6))
sns.scatterplot(x='study_hours_per_day', y='productivity_score', hue='cluster', data=df, palette='Set1')
plt.title('Clusters of Student Productivity')
plt.xlabel('Study Hours per Day')
plt.ylabel('Productivity Score')
plt.legend(title='Cluster')
plt.show()
```

The scatter plot visualises the clusters identified by K-means clustering. Each point represents a student, coloured according to cluster membership. A general positive relationship can be observed between study hours per day and productivity score. Cluster 2 appears to contain students with generally higher productivity levels, while Cluster 3 contains students with lower productivity scores. This suggests that different behavioural patterns may be associated with varying levels of productivity.

```python
#calculate the average behaviour for each cluster and show it in a bar plot.
cluster_behavior = df.groupby('cluster')[['study_hours_per_day', 'social_media_hours', 'youtube_hours', 'sleep_hours', 'phone_usage_hours', 'gaming_hours', 'stress_level','breaks_per_day','exercise_minutes']].mean().reset_index()
cluster_behavior_melted = cluster_behavior.melt(id_vars='cluster', var_name='behavior', value_name='average')
plt.figure(figsize=(12, 8))
sns.barplot(x='behavior', y='average', hue='cluster', data=cluster_behavior_melted)
plt.title('Average Behavior by Cluster')
plt.xlabel('Behavior')
plt.ylabel('Average')
plt.legend(title='Cluster')
plt.xticks(rotation=45)
plt.show()
```

The bar chart presents the average behavioural characteristics for each cluster identified through K-means clustering. The clusters display distinct behavioural patterns. For example, Cluster 0 shows higher phone usage and lower exercise levels, while Cluster 2 demonstrates lower screen time and relatively balanced behaviours. Cluster 1 is associated with higher stress levels, whereas Cluster 3 shows higher gaming and exercise activity. These patterns indicate that students can be grouped into different behavioural profiles that may influence their productivity.
