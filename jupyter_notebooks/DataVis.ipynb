# Section 2

Section 2 content

```python
# Load the cleaned dataset using an absolute path

project_root = r'c:\Users\leha_\Documents\vscode-projects\Student-Productivity-and-Behaviours-Project'
df = pd.read_csv(os.path.join(project_root, 'CleanData', 'cleaned_data.csv'))
df.head()
```

Core concepts such as mean, median, standard deviation, hypothesis testing, and basic probability. These will be outlined below.

Mean is the average of a set of numbers. The mean is a very important measure of central tendency, which represents the typical value of a dataset. The mean provides a summary of the entire data making it easier to understand the overall patterns of the dataset. I will use the mean to help understand the average of a student's study_hours_per_day, assignments_completed and final_grade

```python
# calculate the mean study hours per day, attendance percentage, final grade.
mean_study_hours = df['study_hours_per_day'].mean()
mean_attendance_percentage = df['attendance_percentage'].mean()
mean_final_grade = df['final_grade'].mean()
print("Mean study hours per day:", mean_study_hours)
print("Mean attendance percentage:", mean_attendance_percentage)
print("Mean final grade:", mean_final_grade)
```

Median is the middle value of the data in an ascending or descending order. The median is known as the central value of any given data. 

```python
#median of study hours per day, attendance percentage, final grade.
median_study_hours = df['study_hours_per_day'].median()
median_attendance_percentage = df['attendance_percentage'].median()
median_final_grade = df['final_grade'].median()
print("Median study hours per day:", median_study_hours)
print("Median attendance percentage:", median_attendance_percentage)
print("Median final grade:", median_final_grade)
```

Standard deviation describes how spread out the values are around the mean. There is a high standard deviation and low standard deviation. A low standard deviation shows that most of the numbers are close to the mean. A high standard deviation means that the value are spread out over a wider range. This can provide how consistent student behaviour.

```python
# standard deviation of study hours per day, attendance percentage, final grade.
std_study_hours = df['study_hours_per_day'].std()
std_attendance_percentage = df['attendance_percentage'].std()
std_final_grade = df['final_grade'].std()
print("Standard deviation of study hours per day:", std_study_hours)
print("Standard deviation of attendance percentage:", std_attendance_percentage)
print("Standard deviation of final grade:", std_final_grade)
```

Hypothesis testing compares two conflicting hypothesis. the key concept is Null Hypothesis which assumes that there is no effect.
The Alternative Hypothesis which shows that there is a difference or a significant effect. 
P-value is to check if the data of the null hypothesis is true. If it is under 0.05 then the null hypothesis is false.
T-test is used to compare the averages of two groups to see if they are significantly different from each other. We will use this to compare the mean final grade of students of group 1 who study mpre than 4 hours per day and group 2 who study less than 4 hours per day

```python
#hypothesis testing using t-test to compare the mean final grade of students who study more than 4 hours per day and those who study less than 4 hours per day.
from scipy.stats import ttest_ind
group1 = df[df['study_hours_per_day'] > 4]['final_grade']
group2 = df[df['study_hours_per_day'] <= 4]['final_grade']
t_stat, p_value = ttest_ind(group1, group2)
print("T-statistic:", t_stat)
print("P-value:", p_value)
```

The p-value is under 0.5 so the null hypothesis is false.

Basic Probability confirms the likelihood of something is to happen. It gives a numerical value to help solve uncertainty in a dataset. It is expressed as a number between 0 and 1. 0 meaning impossible and 1 is certain.

```python
#probability of students who study more than 4 hours per day getting a final grade above 85.
probability = len(df[(df['study_hours_per_day'] > 4) & (df['final_grade'] > 85)]) / len(df[df['study_hours_per_day'] > 4])
print("Probability of students who study more than 4 hours per day getting a final grade above 85:", probability)
```
